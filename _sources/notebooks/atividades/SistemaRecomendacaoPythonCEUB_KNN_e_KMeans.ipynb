IMPLEMENTAÇÃO DE CLUSTERING


```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
#Normalizando os dados para clustering
scaler = StandardScaler()
filmes_scaled = scaler.fit_transform(filmes_pivot)
```

```python
#Aplicar KMeans
n_clusters = 8
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
filmes_pivot['Cluster'] = kmeans.fit_predict(filmes_scaled)
```

```python
filmes_pivot.head(20)
```

```python
#Função para imprimir os clusters

def imprimir_clusters(df):
    for cluster in range(n_clusters):
        print(f"\nCluster {cluster}:")
        filmes_no_cluster = df[df['Cluster'] == cluster].index.tolist()
        for filme in filmes_no_cluster:
            print(f"- {filme}")
```

```python
#Imprimir os filmes de cada cluster
imprimir_clusters(filmes_pivot)
```

```python
#Função de recomendação baseada em KMeans que também informa o cluster do filme

def recomendar_filme(filme_input,df):
    #Encontrar o cluster do filme entrada
    cluster_filme = df[df.index == filme_input]['Cluster'].values[0]

    #Filtrar os filmes do mesmo cluster
    recomendacoes = df[df['Cluster'] == cluster_filme].index.tolist()
    recomendacoes.remove(filme_input) #Remover o próprio filme da recomendação

    print(f"Filme de entrada: '{filme_input}' pertence ao Cluster {cluster_filme}")
    print(f"Recomendações para quem gostou de '{filme_input}':")

    for filme in recomendacoes:
        cluster_recomendado = df[df.index == filme]['Cluster'].values[0]
        print(f"- {filme} (Cluster {cluster_recomendado})")
```

```python
#Testar a função de recomendação
filme_teste = 'Titanic' #Exemplo de filme
recomendar_filme(filme_teste,filmes_pivot)
```

USANDO DISTÂNCIA EUCLIDIANA

```python
#Função para calcular a distância euclidiana
def distancia_euclidiana(vetor1,vetor2):
    return np.sqrt(np.sum((vetor1 - vetor2)**2))
```

```python
#Função de recomendação com KMeans e Distância Euclidiana

def recomendar_filmes_kmeans_euclidiana(filme_input,df,n_recomendacoes=5):
    #Encontrar o cluster do filme de entrada
    cluster_filme = df[df.index == filme_input]['Cluster'].values[0]

    #Filtrar os filmes do mesmo cluster
    filmes_no_cluster = df[df['Cluster'] == cluster_filme]

    #Remover o filme deentrada do cluster para não recomendá-lo
    filmes_no_cluster = filmes_no_cluster.drop(filme_input,axis=0)

    #Lista par armazenar as distâncias calculadas
    distancias = []

    #Vetor do filme de entrada (sem a coluna do cluster)
    vetor_filme_input = df.loc[filme_input].drop(['Cluster']).values

    #Calcular a distancia euclidiana entre o filme de entrada e os demais filmes do cluster
    for filme in filmes_no_cluster.index:
        vetor_filme = filmes_no_cluster.loc[filme].drop(['Cluster']).values
        distancia = distancia_euclidiana(vetor_filme_input,vetor_filme)
        distancias.append((filme, distancia))

    #Ordenar pela menor distancia (mais proximo)
    distancias_ordenadas = sorted(distancias, key=lambda x: x[1])

    #Exibir as recomendações
    print(f"Filme de entrada: '{filme_input}' pertence ao cluster '{cluster_filme}'")
    print(f"Recomendacoes para quem gostou de '{filme_input}':")

    for i in range(n_recomendacoes):
        filme_recomendado, dist = distancias_ordenadas[i]
        print(f" - {filme_recomendado} (Distância Euclidiana: {dist:.2f})")
```

```python
#Testar a função de recomendação
filme_teste = 'Toy Story' #Exemplo de teste
recomendar_filmes_kmeans_euclidiana(filme_teste,filmes_pivot)
```

USANDO COMBINAÇÃO KMEANS + KNN

```python
#Função de recomendação com KMeans e KNN dentro do cluster

def recomendar_filmes_kmeans_knn(filme_input,df,n_recomendacoes):
    #Encontrar o cluster do filme de entrada
    cluster_filme = df[df.index == filme_input]['Cluster'].values[0]

    #Filtrar os filmes do mesmo cluster
    filmes_no_cluster = df[df['Cluster'] == cluster_filme]

    #Remover o filme de entrada do cluster para não recomendá-lo
    filmes_no_cluster = filmes_no_cluster.drop(filme_input,axis=0)

    #Criar um modelo KNN para calcular a similaridade dentro do cluster
    knn = NearestNeighbors(n_neighbors=n_recomendacoes, algorithm='brute')
    knn.fit(filmes_no_cluster.drop(columns=['Cluster'])) #usar os dados sem a coluna de cluster

    #Encontrar as recomendações mais próximas dentro do cluster
    distances, sugestions = knn.kneighbors(df.loc[filme_input].drop(['Cluster']).values.reshape(1,-1))

    #Exibir as recomendações
    print(f"Filme de entrada: '{filme_input}' pertence ao cluster '{cluster_filme}'")
    print(f"Recomendacoes para quem gostou de '{filme_input}':")

    for i in range(len(sugestions[0])):
        filme_recomendado = filmes_no_cluster.index[sugestions[0][i]]
        cluster_recomendado = filmes_no_cluster.loc[filme_recomendado]['Cluster']
        print(f" - {filme_recomendado} (Cluster {cluster_recomendado})")
```

```python
#Testar a função de recomendação
filme_teste = 'The Matrix' #Exemplo de filme
recomendar_filmes_kmeans_knn(filme_teste,filmes_pivot,5)
```
